```python
class Product:
    def __init__(self, product_name, unit_price):
        self.product_name = product_name
        self.unit_price = unit_price

    def total_cost(self, quantity):
        total = self.unit_price * quantity
        return total

Notebook = Product("Notebook", 30)
print(Notebook.total_cost(5))
```

```python
def text_to_float(price_entry):
    price_entry = float(price_entry)
    if price_entry < 0:
        raise ValueError(f"The price of the product must be positive, not {price_entry}. Enter again")
    else:
        print(f"Price {price_entry}, successfully entered.")

text_to_float("90")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

sales = pd.read_csv("sales.csv")
sales.dropna(inplace = True)
sales.drop_duplicates(inplace = True)

product_a = sales[sales.Product["A"]]
plt.figure(figsize = (10, 5))
plt.plot(product_a.Day, product_a.Sales)
plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Sales against Day")
plt.show()
```

```python
import numpy as np
a = np.array([[4, 2], [8, 12]])
b = np.array([20, 72])

x = np.linalg.solve(a, b)
print(x)
```

```python
import numpy as np
from scipy.linalg import lu

A = np.array([[4, 2], [8, 12]])
p, l, u = lu(A)
print(f"p = \n {p}")
print(f"l = \n {l}")
print(f"u = \n {u}")
```

```python
import numpy as np
A = np.array([[2, 1], [1, 2]])
eigenvalue, eigenvector = np.linalg.eig(A)

print("Eigenvalue: ", eigenvalue)
print("Eigenvector: \n", eigenvector)

lam = eigenvalue[0]
v = eigenvector[:, 0]

print("A @ v (Av) = ", A @ v)
print("Lambda * v = ", lam * v)
```

```python
from scipy.integrate import quad

def f(x):
    return x**3 + 2*x + 5

area, error = quad(f, 1, 2)
print("Area =", area)      # 11.25
print("Error estimate =", error)
```
