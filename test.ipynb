```python
class Vehicle:
    def __init__(self, name, speed):
        self.name = name
        self.speed = speed
    
    def distance(self, hours):
        return self.speed * hours

v1 = Vehicle("V1", 60)
v1.distance(2)
```

```python
def load_delays(path):
    delays = []
    try:
        with open(path, "r") as file:
            for line in file:
                delays.append(float(line))
    except FileNotFoundError:
        print(f"File not found at {path}. Check file name and folder.")
        return []
    
    return delays
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("deliveries.csv")
df['Route'] = df['Route'].str.strip()
df = df[df['DelayMinutes'].dropna(inplace = True)]
df = df[df['DelayMinutes'] >= 0]

North = df[df['Route'] == 'North']

plt.figure(figsize = (10, 5))
plt.plot(North.Day, North.DelayMinutes)
plt.xlabel('Day')
plt.ylabel('DelayMinutes')
plt.title("A line chart showing delays for the North route")
plt.show()
```
